# Pilot RAG Analysis
Đây là file phân tích mẫu (Histogram + Descriptive stats) cho kết quả chạy Pilot.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc kết quả Pilot
try:
    df_pilot = pd.read_csv('pilot_llm_output.csv')
    display(df_pilot.head())
except FileNotFoundError:
    print("Chưa tìm thấy file pilot_llm_output.csv")
```

```python
# Thống kê mô tả (Descriptive stats)
if 'df_pilot' in locals():
    # Tính độ dài của context gốc và context bị đột biến
    df_pilot['original_length'] = df_pilot['original_context'].fillna('').apply(len)
    df_pilot['mutated_length'] = df_pilot['mutated_context'].fillna('').apply(len)
    
    display(df_pilot[['original_length', 'mutated_length']].describe())
```

```python
# Vẽ Histogram phân bố độ dài
if 'df_pilot' in locals():
    plt.figure(figsize=(10, 5))
    sns.histplot(df_pilot['original_length'], color='blue', label='Original Context', kde=True, alpha=0.5)
    sns.histplot(df_pilot['mutated_length'], color='red', label='Mutated Context', kde=True, alpha=0.5)
    plt.title('Phân bố độ dài Context (Pilot)')
    plt.xlabel('Số ký tự')
    plt.ylabel('Tần suất')
    plt.legend()
    plt.show()
```
